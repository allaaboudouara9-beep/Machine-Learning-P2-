```python
import warnings
warnings.filterwarnings('ignore') # We can suppress the warnings

# importing the dataset with pandas
df= pd.read_csv('AAPL_historical_data.csv')

```

```python
df
```

```python
print(df.head())
print(df.columns)
print(df.info())
```

```python
df = df.drop([0, 1])
```

```python
df.head()
```

```python
df.columns = ['Date', 'Close', 'High', 'Low', 'Open', 'Volume']
```

```python
#Convert columns to numeric
for col in ['Close', 'High', 'Low', 'Open', 'Volume']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df = df.dropna()
```

```python
X = df[['Close', 'High', 'Low', 'Open', 'Volume']]
```

```python
print(X.shape)
print(X.head())
```

```python
print(df.select_dtypes(include='number').shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# import KMeans library for clustering
from sklearn.cluster import KMeans

# Applying KMeans to the dataset/ Creating the kmeans classifier
kmeans = KMeans(n_clusters = 2, max_iter = 300, n_init = 10, random_state = 38)

# n_initial, default = 10, Number of time the k-means algorithm will be run with different centroid seeds. 
# The final results will be the best output of n_init consecutive runs in terms of inertia.
y_kmeans = kmeans.fit_predict(X_scaled)
```

```python
# Visualising the clusters using scatter() method

plt.scatter(X_scaled[y_kmeans == 0, 0], X_scaled[y_kmeans == 0, 1], s = 50, c = 'red', label = 'Cluster 0')
plt.scatter(X_scaled[y_kmeans == 1, 0], X_scaled[y_kmeans == 1, 1], s = 50, c = 'blue', label = 'Cluster 1')
plt.scatter(X_scaled[y_kmeans == 2, 0], X_scaled[y_kmeans == 2, 1], s = 50, c = 'green', label = 'Cluster 2')

# Plotting the centroids of the clusters
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:,1], s = 150, c = 'yellow', label = '')

plt.title("KMeans Clusters -  Stock Data")
plt.xlabel("Feature 1 (Scaled)")
plt.ylabel("Feature 2 (Scaled)")
plt.legend()
plt.show()

# A legend is an area describing the elements of the graph. In the matplotlib library, there's a function called legend() 
# which is used to Place a legend on the axes.
plt.legend( loc ="best") 
```

Elbow Method 

```python
# Finding the optimum number of clusters for k-means classification
from sklearn.cluster import KMeans
ssw = []                  # Declare an array for the storage of values for Sum of Squares (SSW) Within Cluster

# Set the loop from the minimum and maximum values
for i in range(1, 11):
    kmeans = KMeans(n_clusters = i, max_iter = 300, n_init = 10, random_state = 38)  # Initialise an object kmeans by calling 
                                                                                     # an object KMeans()
    kmeans.fit(X_scaled)                                                                    # Training the model by calling fit() method
    ssw.append(kmeans.inertia_)                                                      # Store kmeans.inertia_ in the array ssw

# Plotting the results onto a line graph with points in a different color
plt.plot(range(1, 11), ssw, linestyle='-', color='blue', label='SSW line')
plt.scatter(range(1, 11), ssw, color='red', marker='o', s=50, label='SSW points')

plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('SSW')  # within cluster sum of squares
plt.legend()
plt.show()
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

ssw = []

# Try k = 1 to 4
for i in range(1, 5):
    kmeans = KMeans(n_clusters=i, max_iter=300, n_init=10, random_state=38)
    kmeans.fit(X_scaled)   # use scaled data
    ssw.append(kmeans.inertia_)

# Plot
plt.figure(figsize=(8,5))

plt.plot(range(1, 5), ssw, linestyle='-', color='blue')
plt.scatter(range(1, 5), ssw, color='red')

plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('SSW (Inertia)')
plt.show()
```

 Silhouette Score 

```python
# import silhouette score from sklearn library
from sklearn.metrics import silhouette_score

# Display the silhouette score values
print(f'Silhouette Score(n = 3): {silhouette_score(X_scaled, y_kmeans)}')
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Loop through cluster numbers from 2 to 9
for n_clusters in range(2, 10):
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    labels = kmeans.fit_predict(X_scaled)   # Fit and predict cluster labels
    score = silhouette_score(X_scaled, labels)
    print(f"Silhouette Score (n = {n_clusters}): {score:.4f}")
```

Davies–Bouldin Index

```python
from sklearn.metrics import davies_bouldin_score
from sklearn.cluster import AgglomerativeClustering

# --- KMeans ---
db_km = davies_bouldin_score(X_scaled, y_kmeans)

# --- Agglomerative ---
agg = AgglomerativeClustering(n_clusters=2)
labels_agg = agg.fit_predict(X_scaled)

db_agg = davies_bouldin_score(X_scaled, labels_agg)

# Print results
print("KMeans Davies-Bouldin Index:", db_km)
print("Agglomerative Davies-Bouldin Index:", db_agg)
```

```python
from sklearn.metrics import silhouette_score

sil_km = silhouette_score(X_scaled, y_kmeans)
sil_agg = silhouette_score(X_scaled, labels_agg)

print("KMeans Silhouette:", sil_km)
print("Agglomerative Silhouette:", sil_agg)
```
